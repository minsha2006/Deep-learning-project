```python
import torch

tensor = torch.randn(5, 4)

print("Original Tensor (5x4):\n", tensor)

mean_val = tensor.mean()
max_val = tensor.max()
std_val = tensor.std()

print("\nMean:", mean_val.item())
print("Max:", max_val.item())
print("Standard Deviation:", std_val.item())

reshaped_tensor = tensor.view(4, 5)

print("\nReshaped Tensor (4x5):\n", reshaped_tensor)
```
